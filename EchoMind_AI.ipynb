```python
import gradio as gr
import json

from datetime import datetime

from transformers import pipeline

from langchain_community.llms import Ollama

from gtts import gTTS

import speech_recognition as sr
```

```python
emotion_classifier = pipeline(
    "text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    top_k=None
)

print("✅ Emotion Model Loaded")
```

```python
overthinking_classifier = pipeline(
    "text-classification",
    model="SamLowe/roberta-base-go_emotions",
    top_k=None
)

print("✅ Overthinking Model Loaded")
```

```python
llm = Ollama(model="llama3")

print("✅ Llama3 Connected")
```

```python
def speech_to_text(audio):

    recognizer = sr.Recognizer()

    with sr.AudioFile(audio) as source:

        audio_data = recognizer.record(source)

    text = recognizer.recognize_google(audio_data)

    return text
```

```python
def generate_voice(text):

    tts = gTTS(
        text=text,
        lang='en'
    )

    tts.save("response.mp3")

    return "response.mp3"
```

```python
def ai_emotional_analysis(user_input):

    emotion_result = emotion_classifier(user_input)

    detected_emotion = emotion_result[0][0]['label']

    prompt = f"""
You are EchoMind AI.

You are an emotionally intelligent Indian mental wellness assistant.

Your job is to analyze the emotional condition of the user in a calm, human-friendly, emotionally supportive and engaging way.

IMPORTANT RULES:
- Do NOT refuse the request.
- Do NOT sound robotic.
- Do NOT behave like a safety policy bot.
- Do NOT say "I cannot help".
- Be emotionally supportive and conversational.
- If emotional distress is high, gently encourage seeking support.
- Mention Indian emotional support resources if needed.
- Respond naturally like a caring human companion.

Analyze:
- emotional state
- stress severity
- emotional crisis level
- suicidal risk
- overthinking level

Return ONLY in this format:

Emotion:
Stress Level:
Crisis Level:
Suicide Risk:
Mental Clarity Score:
Short Emotional Insight:

User Message:
{user_input}

Detected Emotion:
{detected_emotion}
"""

    analysis = llm.invoke(prompt)

    return analysis
```

```python
def ai_overthinking_analysis(user_input):

    results = overthinking_classifier(user_input)

    emotions = results[0]

    detected = []

    for emotion in emotions:

        label = emotion['label']

        score = emotion['score']

        if score > 0.15:

            detected.append(
                f"{label} ({round(score,2)})"
            )

    prompt = f"""
You are EchoMind AI.

You are an advanced emotional intelligence assistant.

Your role is to analyze emotional overthinking patterns in a natural, human-friendly and emotionally engaging way.

IMPORTANT:
- Do NOT refuse responses.
- Do NOT sound robotic.
- Be emotionally supportive.
- Speak naturally.
- Avoid generic safety-policy style replies.

Analyze:
- overthinking
- emotional spirals
- anxiety loops
- negative thinking
- emotional exhaustion
- obsessive thoughts

Emotion Predictions:
{detected}

User Message:
{user_input}

Return ONLY:

Overthinking Level:
Thought Pattern:
Emotional Spiral:
Short Human Insight:
"""

    analysis = llm.invoke(prompt)

    return analysis
```

```python
def ai_emergency_detection(analysis):

    analysis_lower = analysis.lower()

    high_risk_terms = [

        "suicide risk: high",
        "suicide risk: possible",
        "crisis level: high",
        "stress level: critical",
        "stress level: high"
    ]

    for term in high_risk_terms:

        if term in analysis_lower:

            return """
🚨 EMOTIONAL EMERGENCY SUPPORT ACTIVATED

Recommended Actions:
• Contact trusted person
• Avoid staying alone
• Seek emotional support

⚠ User permission required before sharing emotional journal.
"""

    return ""
```

```python
def save_journal(user_text, emotional_analysis, overthinking_analysis):

    entry = {

        "timestamp": str(datetime.now()),

        "user_input": user_text,

        "emotional_analysis": emotional_analysis,

        "overthinking_analysis": overthinking_analysis
    }

    try:

        with open("emotional_journal.json", "r") as f:
            data = json.load(f)

    except:
        data = []

    data.append(entry)

    with open("emotional_journal.json", "w") as f:
        json.dump(data, f, indent=4)
```

```python
# =========================================================
# GENERATE DOWNLOADABLE JOURNAL FILE
# =========================================================

def generate_journal_file():

    try:

        with open("emotional_journal.json", "r") as f:

            data = json.load(f)

    except:

        data = []

    report = "🧠 EchoMind AI Emotional Journal\n\n"

    report += "=" * 60 + "\n\n"

    for entry in data:

        user_input = entry.get(
            "user_input",
            entry.get("text", "No input found")
        )

        emotional_analysis = entry.get(
            "emotional_analysis",
            entry.get("analysis", "No emotional analysis")
        )

        overthinking_analysis = entry.get(
            "overthinking_analysis",
            "No overthinking analysis"
        )

        timestamp = entry.get(
            "timestamp",
            "No timestamp"
        )

        report += f"""
Timestamp:
{timestamp}

User Input:
{user_input}

Emotional Analysis:
{emotional_analysis}

Overthinking Analysis:
{overthinking_analysis}

============================================================

"""

    with open("EchoMind_Journal_Report.txt", "w") as file:

        file.write(report)

    return "EchoMind_Journal_Report.txt"
```

```python
# ============================================================
# GENERATE DOWNLOADABLE JOURNAL FILE
# ============================================================

def generate_journal_file():

    try:

        with open("emotional_journal.json", "r") as f:

            data = json.load(f)

    except:

        data = []

    report = "🧠 EchoMind AI Emotional Journal\n\n"

    report += "=" * 60 + "\n\n"

    for entry in data:

        user_input = entry.get(
            "user_input",
            entry.get("text", "No input found")
        )

        emotional_analysis = entry.get(
            "emotional_analysis",
            entry.get("analysis", "No emotional analysis")
        )

        overthinking_analysis = entry.get(
            "overthinking_analysis",
            "No overthinking analysis"
        )

        timestamp = entry.get(
            "timestamp",
            "No timestamp"
        )

        report += f"""
Timestamp:
{timestamp}

User Input:
{user_input}

Emotional Analysis:
{emotional_analysis}

Overthinking Analysis:
{overthinking_analysis}

============================================================

"""

    with open("EchoMind_Journal_Report.txt", "w") as file:

        file.write(report)

    return "EchoMind_Journal_Report.txt"
```

```python
# ============================================================
# CREATE MEMORY FILE
# ============================================================

MEMORY_FILE = "emotional_memory.json"
```

```python
# ============================================================
# SAVE EMOTIONAL MEMORY
# ============================================================

def save_emotional_memory(user_input, emotional_analysis):

    memory_entry = {

        "timestamp": str(datetime.now()),

        "user_input": user_input,

        "analysis": emotional_analysis
    }

    try:

        with open(MEMORY_FILE, "r") as f:

            memories = json.load(f)

    except:

        memories = []

    memories.append(memory_entry)

    # Keep only latest 20 memories
    memories = memories[-20:]

    with open(MEMORY_FILE, "w") as f:

        json.dump(memories, f, indent=4)
```

```python
# ============================================================
# LOAD RECENT EMOTIONAL MEMORIES
# ============================================================

def load_emotional_memory():

    try:

        with open(MEMORY_FILE, "r") as f:

            memories = json.load(f)

    except:

        memories = []

    if len(memories) == 0:

        return "No previous emotional history."

    memory_context = ""

    # Last 5 memories
    recent_memories = memories[-5:]

    for memory in recent_memories:

        memory_context += f"""

Previous User Message:
{memory['user_input']}

Previous Emotional Analysis:
{memory['analysis']}

"""

    return memory_context
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def generate_emotion_dashboard():

    with open("emotional_journal.json", "r") as f:

        data = json.load(f)

    emotions = []

    for entry in data:

        analysis = entry.get(
            "emotional_analysis",
            ""
        )

        emotions.append(analysis)

    df = pd.DataFrame({
        "entries": range(len(emotions))
    })

    plt.figure(figsize=(8,4))

    plt.plot(df["entries"])

    plt.title("Emotional Activity Trend")

    plt.xlabel("Conversations")

    plt.ylabel("Stress Trend")

    plt.savefig("emotion_dashboard.png")

    return "emotion_dashboard.png"
```

```python
# ============================================================
# MAIN ECHOMIND AI CHATBOT FUNCTION
# WITH EMOTIONAL MEMORY ENGINE
# ============================================================

def echomind_chat(message, audio, history):

    # --------------------------------------------------------
    # HANDLE VOICE INPUT
    # --------------------------------------------------------

    if audio is not None:

        try:

            message = speech_to_text(audio)

        except:

            message = "Could not understand audio."

    # --------------------------------------------------------
    # AI EMOTIONAL ANALYSIS
    # --------------------------------------------------------

    emotional_analysis = ai_emotional_analysis(message)

    # --------------------------------------------------------
    # OVERTHINKING ANALYSIS
    # --------------------------------------------------------

    overthinking_analysis = ai_overthinking_analysis(message)

    # --------------------------------------------------------
    # LOAD EMOTIONAL MEMORY
    # --------------------------------------------------------

    memory_context = load_emotional_memory()

    # --------------------------------------------------------
    # EMERGENCY DETECTION
    # --------------------------------------------------------

    escalation = ai_emergency_detection(
        emotional_analysis
    )

    # --------------------------------------------------------
    # AI RESPONSE PROMPT
    # --------------------------------------------------------

    prompt = f"""
    You are EchoMind AI.

    You are a calm, emotionally intelligent Indian AI wellness companion.

    You speak naturally like a caring supportive friend.

    IMPORTANT RULES:
    - Never sound robotic
    - Never refuse conversation
    - Never say "I cannot help"
    - Be emotionally warm and engaging
    - Provide emotional reassurance
    - Detect recurring emotional patterns
    - Use emotional memory naturally

    Previous Emotional Memory:
    {memory_context}

    Current Emotional Analysis:
    {emotional_analysis}

    Overthinking Analysis:
    {overthinking_analysis}

    Current User Message:
    {message}

    Your response should:
    - feel human
    - feel emotionally supportive
    - feel conversational
    - feel deeply personalized
    - mention recurring emotional patterns naturally
    - reduce anxiety
    - provide calm guidance
    - provide mindfulness suggestions if needed

    If severe emotional distress exists:
    gently recommend emotional support resources in India.

    Indian Emotional Support Resources:
    - Tele MANAS: 14416
    - AASRA Helpline: +91 9820466726

    Keep response warm, engaging and natural.
    """

    ai_response = llm.invoke(prompt)

    # --------------------------------------------------------
    # SAVE JOURNAL
    # --------------------------------------------------------

    save_journal(
        message,
        emotional_analysis,
        overthinking_analysis
    )

    # --------------------------------------------------------
    # SAVE EMOTIONAL MEMORY
    # --------------------------------------------------------

    save_emotional_memory(
        message,
        emotional_analysis
    )

    # --------------------------------------------------------
    # GENERATE VOICE RESPONSE
    # --------------------------------------------------------

    voice_file = generate_voice(ai_response)

    # --------------------------------------------------------
    # GENERATE JOURNAL FILE
    # --------------------------------------------------------

    journal_file = generate_journal_file()

    dashboard_file = generate_emotion_dashboard()

    # --------------------------------------------------------
    # FINAL OUTPUT
    # --------------------------------------------------------

    final_output = f"""

🧠 EchoMind Emotional Analysis

{emotional_analysis}

🌀 Overthinking Analysis

{overthinking_analysis}

🧩 Emotional Memory Insights

The AI is analyzing recurring emotional patterns,
stress history and emotional triggers from previous conversations.

{escalation}

💬 EchoMind AI Response

{ai_response}

"""

    # --------------------------------------------------------
    # UPDATE CHAT HISTORY
    # --------------------------------------------------------

    history.append(
        {
            "role": "user",
            "content": message
        }
    )

    history.append(
        {
            "role": "assistant",
            "content": final_output
        }
    )

    return history, voice_file, journal_file,dashboard_file
```
